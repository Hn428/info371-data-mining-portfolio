### 1. Introduction
---
*(Introduce the project and describe the objectives. This section will provide an overview of the entire project including the description of the data sets and the specific data analytics methods and techniques the team will use for analyzing the data to address the problem.)*

Daily smartphone use is a behavioral indicator that impacts the quality of sleep, levels of stress, and academic/work productivity, whether positive or negative. In this project, we take a dataset with 50,000 records of analytics on user lifestyle and create both descriptive and predictive models to analyze associations of phone usage, screen time on social media, duration of sleep, caffeine consumption and productivity as well as stress levels.

The objectives are as follows:
1. Regression: From behavioral and demographic characteristics, predict Work_Productivity_Score (1–10).
2. Classification: Identify higher risk of stress by recording Stress Level (1–10) into groups (ex. low/middle/high or high vs not-high).
3. Clustering: Using unsupervised learning, find user “lifestyle personas” and describe interpreted clusters by usage, sleep, stress, and productivity.

Techniques will be applied to preprocessing of categorical variables, engineering of features (derived metrics and interaction features), selection of features, cross-validation, model evaluation and comparison (ensemble methods and baseline), and relevant metrics (appropriate) will be used for evaluation.

**Evaluation Metrics:**
- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)

## Task B: Stress Risk Classification

Transform y_i^stress ∈ [1,10] into categorical labels.

### Option 1: Binary Classification
- High stress = 1 if stress ≥ 8

- Low: 1–4
- Medium: 5–7
- High: 8–10

**Goal:**  
Maximize classification performance.

**Evaluation Metrics:**
- F1-score
- ROC-AUC (for binary case)

**Validation Strategy:**
- Cross-validation

### 4. Other Things you Would Like to Describe
---
**Planned feature engineering**

We will engineer derived features such as:

Sleep deficit: max(0, 7 - Sleep_Hours)

*   Social media ratio: Social_media_Hours / Daily_Phone_Hours
*   Rest to screen ratio: Sleep_Hours / Daily_Phone_Hours
*   App intenaity: app_usage / Daily_Phone_Hours
*   Weekend Amplification: Weekend_Screen_Time_Hours / Daily_Phone_Hours
*   Social Media Interaction: Social_Media_Hours * Sleep_Deficit
*   Caffeine Sleep deficit: Caffeine_Intake_Cups * Sleep_Deficit
*   Cognitive Load Index: (App_Usage_Count * Social_Media_Hours) / Sleep_Hours
*   Usage intensity: Daily_Phone_Hours / (App_Usage_Count + 1)
*   Sleep_Deviation_From_Optimal: |7.5 - Sleep_Hours|
