## Question 3 [60 marks] Use the data set you created in Question 2:
1. Import KNeighborsClassifier from Scikit-Learn
2. Import accuracy_score from Scikit-Learn
3. Extract the two numerical attributes as training data X and the categorical attribute as target y
4. Fit a K Nearest Neighbor classifier on X and y
5. Predict the target labels of the training data X
6. Compute and show the accurcy score of the KNN classifier on the training data X
7. By default, how many nearest neighbors are used in the model?
8. Use different number of nearest neighbors; test K = 1, 2, 3 and show the accuracy results.
9. Does the value of K affect the performance? Why or Why not?

```python
# Extract features (X) and target (y)
X = df[["HoursStudied", "ExamScore"]]
y = df["Passed"]

# -----------------------------
# KNN with default parameters
# -----------------------------

# Create KNN classifier (default K = 5)
knn_default = KNeighborsClassifier()

# Fit model
knn_default.fit(X, y)

# Predict on training data
y_pred_default = knn_default.predict(X)

# Compute accuracy
accuracy_default = accuracy_score(y, y_pred_default)
print("Accuracy with default K:", accuracy_default)

# -----------------------------
# KNN with different values of K
# -----------------------------

for k in [1, 2, 3]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X, y)
    y_pred = knn.predict(X)
    acc = accuracy_score(y, y_pred)
    print(f"Accuracy with K={k}: {acc}")

#results:
#By default, the KNN classifier uses 5 nearest neighbors
#The model was then trained using K values of 1, 2, and 3. The results showed that the value of K affects performance
#smaller K values tend to fit the training data more closely while larger K values provide smoother decision boundaries. This demonstrates the bias-variance tradeoff inherent in the KNN algorithm.
```
